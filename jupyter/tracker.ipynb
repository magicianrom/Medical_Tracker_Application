### Adding new details

```python
id = len(medicine) +1
name = input("Enter medicine's name : ")
ing = input("Whats the ingredient : ")
cata = input("Whats the category : ")
sym = input("whats the symptoms : ")
inst = input("whats the instructions : ")
dos = input("whats the dosage : ")
frq = input("whats the frequency hours: ")
saf = input("whats the saftey precautions :")
urg = input("whats the urgency : ")
```

```python
print(id)
print(name)
print(ing)
print(cata)
print(sym)
print(inst)
print(dos)
print(frq)
print(saf)
print(urg)
```

```python
updated_data = [{"id": id, "name": name, "ingredient":ing, "category":cata, "symptoms": sym,
"instructions":inst, "dosage": dos, "frequency_hours":frq, "safety" :saf, "urgency": urg,
}]
```

```python
df = pd.DataFrame(updated_data)
```

```python
medicine = pd.concat([medicine, df])
```

```python
medicine.to_csv("../../excercises/week2/Medical_Tracker.csv")
```

```python
medicine = pd.read_csv("Medical_Tracker.csv")

medicine_name = input("Which medicine do you want to add? ")

selected = medicine[medicine["name"] == medicine_name].loc[0]
print(selected)
```

```python
name = selected["name"]
ing = selected["ingredient"]
dosage = selected["dosage"]
frequency = selected["frequency_hours"]
```

```python
print(name)
print(ing)
print(dosage)
print(frequency)
```

```python
exp = input("whats the expiry date")

taken_medicine = input("Have you taken the medicine")

if taken_medicine == "yes" or "Yes":
    now = datetime.now()
    next = now + timedelta(hours=int(frequency))

print("medicine taken!")
print(f"next dosage: {next}")
```

```python
print(exp)
print(f'the next dosage is after {next} hours')
```

```python
client = [{"name": name, "ingredient":ing,"dosage": dosage, "expiry":exp, "next_dosage": next,
}]
```

```python
df = pd.DataFrame(client)
```
